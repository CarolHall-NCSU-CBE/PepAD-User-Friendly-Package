```python
length=7

i = 1
chainids=[5,6,7,8,17,18,19,20]
 
mod_lines=[]
file_path = "pep_10ns.pdb"

with open(file_path, 'r') as f:
    lines = f.readlines()   

j=1 # atom counter
for line in lines:
    if line.startswith("ATOM"):
        anum = line[4:11]
        amino_name = line[17:20]
        resid = line[22:26]

        if (int(resid)%length==1):
            line = line.replace(amino_name+" ", "N"+amino_name)
        elif (int(resid)%length==0):
            line = line.replace(amino_name+" ", "C"+amino_name)

        k=0
        for chainid in chainids:

            if (int((int(resid)-1)/length)+1) == chainid:
                anum_new = f"{j:7d}"
                n_chain_before=chainid - 1 - k
                resid_new = f"{(int(resid)-n_chain_before*length):4d}"
                j+=1
                line = line[:21] + " "+ resid_new + " " + line[27:]
                line = line[:4] + anum_new + " " + line[12:]

                mod_lines.append(line)
            k+=1

with open("comp1.pdb", 'w') as f:
    for line in mod_lines:
        f.write(line)

i+=1
```
