```python
length=16

pep="Pep-3"
chainids=[6,7,8,9,17,18,19,20]

mod_lines=[]
file_path = r'{}_25ns.pdb'.format(pep)

with open(file_path, 'r') as f:
    lines = f.readlines()   

j=1 # atom counter
for line in lines:
    if line.startswith("ATOM"):
        anum = line[4:11]
        amino_name = line[17:20]
        resid = line[22:26]
        
        k=0
        for chainid in chainids:
            
            if (int((int(resid)-1)/length)+1) == chainid:
                anum_new = f"{j:7d}"
                n_chain_before=chainid - 1 - k
                resid_new = f"{(int(resid)-n_chain_before*length):4d}"
                j+=1
                line = line[:21] + " "+ resid_new + " " + line[27:]
                line = line[:4] + anum_new + " " + line[12:]

                mod_lines.append(line)
            k+=1
            
with open("comp2 (not aligned).pdb", 'w') as f:
    for line in mod_lines:
        f.write(line)
        
```
